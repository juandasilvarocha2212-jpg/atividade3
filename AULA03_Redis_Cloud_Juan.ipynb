## Passo 1 — Criar o contador zerado

Comando equivalente no Redis:

```redis
SET visitas:home 0
```


````python
resultado = r.set("visitas:home", 0)
print("SET visitas:home 0")
print("Saída:", resultado)
````

## Passo 2 — Simular 5 acessos

O comando `INCR` será executado cinco vezes e, ao final, será feita a leitura do contador com `GET`.


````python
for i in range(5):
    valor = r.incr("visitas:home")
    print(f"INCR visitas:home -> {valor}")

valor_final = r.get("visitas:home")
print("\nGET visitas:home")
print("Valor final:", valor_final)
````

## Passo 3 — Definir expiração de 5 minutos

A chave `visitas:home` receberá um tempo de expiração de 300 segundos.


````python
resultado_expire = r.expire("visitas:home", 300)
ttl = r.ttl("visitas:home")

print("EXPIRE visitas:home 300 ->", resultado_expire)
print("TTL visitas:home ->", ttl)
````

## Passo 4 — Criar cadastro do usuário como Hash

O usuário será armazenado em uma estrutura Hash do Redis.


````python
resultado_hset = r.hset(
    "usuario:1",
    mapping={
        "nome": "Juan da Silva Rocha",
        "email": "seuemail@exemplo.com"
    }
)

nome = r.hget("usuario:1", "nome")

print('HSET usuario:1 nome "Juan da Silva Rocha" email "seuemail@exemplo.com"')
print("Campos adicionados:", resultado_hset)
print("\nHGET usuario:1 nome")
print("Saída:", nome)
````

## Passo 5 — Reflexão final

O `INCR` é útil porque aumenta o valor do contador automaticamente a cada acesso. Já o `EXPIRE` define um tempo para esse contador existir, fazendo com que ele seja apagado automaticamente quando o tempo terminar.
